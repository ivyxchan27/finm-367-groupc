## 1 READING: GMO

This section is not graded, and you do not need to submit your answers. But you are expected to consider these issues and be ready to discuss them.

1. **GMO’s approach.**
   - Why does GMO believe they can more easily predict **long‑run** than **short‑run** asset‑class performance?
   > In the short run the market is a "voting machine": prices are driven by sentiment, flows and overreaction, which GMO says it cannot predict week to week. In the long run it is a "weighing machine": prices revert toward fundamental value. Valuation ratios move slowly and mean-revert slowly, so they say almost nothing about next month's return but a lot about the next 7 years. GMO builds its forecasts on the assumption that P/E multiples and profit margins converge to steady state over 7 years, a horizon "long enough for most market imbalances to work themselves out."

   - What predicting variables does the case mention are used by GMO? Does this fit with the goal of long‑run forecasts?
   > Dividend yield (D/P), the P/E ratio (and the cyclically adjusted P/E), profit margins (E/S) and real sales growth per share. These are backed by macro data: GDP, inflation, corporate profits and recession timing. This fits the long-run goal. These are slow-moving valuation and profitability ratios whose forecasting power builds up over multi-year horizons, and the decomposition $R \approx D/P + \%\Delta(P/E) + \%\Delta(E/S) + \%\Delta S$ only makes sense over a horizon where reversion can happen.

   - How has this approach led to **contrarian** positions?
   > Mean reversion means that whatever has done best (high P/E, high margins) gets the lowest forecast, and whatever has crashed gets the highest. So GMO naturally leans against the crowd: bearish on Japan in the late 1980s, US stocks in the late 1990s and all risky assets in mid-2007, and bullish on US stocks in fall 2008.

   - How does this approach raise **business risk** and **managerial career risk**?
   > Reversion is not guaranteed to be quick or smooth, so a correct call can look wrong for years. Clients judge managers against benchmarks and peers over short windows. Underweighting equities during 1997–2000 cost GMO nearly 60% of its asset-allocation business before the call paid off. Career risk is largest at the asset-class level, where being "wrong about the big picture for two or three years" is visible and costly: "the market can stay irrational longer than the client can stay patient."

2. **The market environment.**
   - We often estimate the market risk premium by looking at a large sample of historic data. What reasons does the case give to be skeptical that the market risk premium will be as high **in the future** as it has been **over the past 50 years**?
   > - In 2012 valuations were still above average (CAPE 19.9 vs. a long-run 16), and profit margins were at records GMO expected to mean-revert.
   > - Some argue that a more stable macro environment and better diversification have lowered required returns and dividend yields.
   > - The "lost decade" (2000–2011) saw stocks trail short-term bonds by 0.1%/yr.
   >
   > GMO's own steady-state equity risk premium (4.2%) is below the 5.2% historical real excess return (1871–2011) and the 6.2% US figure in Exhibit 8.

   - In 2007, GMO forecasts **real excess equity returns** will be negative. What are the biggest drivers of their **pessimistic conditional** forecast relative to the **unconditional** forecast? (See Exhibit 9.)
   > Almost the whole gap comes from expected compression in valuation and margins:
   > - **P/E** reverting from 19.5 to 16.0: **−2.8%/yr** (vs. 0% in steady state)
   > - **Profit margin** reverting from 7.9% to 6.0%: **−3.9%/yr** (vs. 0%)
   > - Sales growth (2.4% vs. 3.2%) and dividend yield (2.3% vs. 2.5%) matter little.
   >
   > Real expected stock return is −1.9% vs. a required 5.7%. The excess return over short-term bonds is −3.9% vs. an equity risk premium of 4.2%.

   - In the 2011 forecast, what components has GMO revised most relative to 2007? Now how does their conditional forecast compare to the unconditional? (See Exhibit 10.)
   > The biggest change is the **P/E** component. Prices fell to a starting P/E of 15.0, and GMO also lowered its steady-state P/E from 16 to 15, so the P/E drag goes from −2.8% to **0%**. The margin drag is essentially unchanged (−3.7%, 7.8% to 6.0%). Expected sales growth rises (2.4% to 2.9%), the dividend-yield contribution rises (2.3% to 2.5%, as the starting yield went from 1.8% to 2.0%), and the expected real short-term bond return falls sharply (2.0% to 0.1%). The conditional real stock return is now **+1.7%** (excess **+1.6%**). That is still well below the unconditional 5.7% (equity risk premium 4.2%), but positive. The remaining pessimism comes almost entirely from expected margin compression.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from hw7utils import summary_stats, lfd, lfd_table, rolling_lfd, period_summary

pd.options.display.float_format = '{:.4f}'.format
```

```python
filename = 'gmo_returns_weekly.xlsx'
TICK_RF = 'USGG3M'

# Load data
gmo = pd.read_excel(filename, sheet_name='gmo', index_col=0, parse_dates=True)
benchmarks = pd.read_excel(filename, sheet_name='benchmarks', index_col=0, parse_dates=True)
indexes = pd.read_excel(filename, sheet_name='indexes', index_col=0, parse_dates=True)
rates = pd.read_excel(filename, sheet_name='rates', index_col=0, parse_dates=True)
info = pd.read_excel(filename, sheet_name='info').set_index('Ticker')


if '_daily' in filename:
    ann_factor = 252
elif '_weekly' in filename:
    ann_factor = 52
elif '_monthly' in filename:
    ann_factor = 12
else:
    # Default to daily if no tag found
    ann_factor = 252
```

### 1. Benchmarks

Construct what we will call the `GBI` (GMWAX Benchmark Index). The `GBI` is constructed as (per detailed in the `info` tab)
* 65% of `MXWD`
* 35% of `LBUSTRUU`

Calculate excess returns of `GMWAX`, `GBI` and of `SPY`.

For all three, estimate the full-sample (annualized)
* mean
* vol
* Sharpe

Also report
* minimum return
* 5th percentile
* maximum drawdown 

Also report the correlation matrix.

```python
#risk-free rate: USGG3M is an annualized yield (SHV is not in the file).
#convert to a weekly rate and lag one week so each week's return uses the yield known at its start.
rf = rates[TICK_RF].shift(1) / ann_factor

#GMWAX Benchmark Index: 65% MSCI ACWI + 35% Bloomberg US Agg (rebalanced weekly)
gbi = (0.65 * indexes['MXWD'] + 0.35 * indexes['LBUSTRUU']).rename('GBI')

rets = pd.concat([gmo, benchmarks, indexes, gbi], axis=1, sort=True)
xrets = rets.sub(rf, axis=0).dropna(how='all')

xsec2 = xrets[['GMWAX', 'GBI', 'SPY']].dropna()
print(f'Sample: {xsec2.index[0].date()} to {xsec2.index[-1].date()}')
display(summary_stats(xsec2, ann_factor))
display(xsec2.corr())
```

> **Risk-free rate.** `SHV` is not in the data file, so I use `USGG3M` (the 3-month T-bill yield), converted to a weekly rate and lagged one week.
>
> Full sample (Nov 1996 - Oct 2025):
>
> | | Mean | Vol | Sharpe | Min | 5th pct | Max DD |

> | GMWAX | 4.9% | 10.4% | 0.47 | −10.6% | −2.2% | −32% |
> | GBI | 5.4% | 11.1% | 0.48 | −13.7% | −2.2% | −41% |
> | SPY | 8.9% | 17.8% | 0.50 | −19.8% | −3.9% | −60% |
>
> - **Risk-adjusted return:** all three have nearly the same Sharpe ratio (~0.47-0.50). GMWAX earns less than either benchmark but with less risk.
> - **Tail risk:** this is where GMWAX stands out. Its worst week, 5th percentile and especially its max drawdown (-32% vs. -41% for GBI and -60% for SPY) are all milder.
> - **Correlations:** GMWAX is 0.86 correlated with the GBI and 0.78 with SPY. The GBI itself is 0.93 correlated with SPY, since 65% of it is global equity.
> - **Compared with the case:** the large Sharpe advantage cited in the case (0.46 vs. 0.17 through 2011) has disappeared over the full sample.

### 2. Attribution

Estimate an attribution LFD of `GMWAX` on
* `SPY`
* `GBI`

That is, estimate a univariate regression for each.

Report the 
* alpha (annualized)
* beta
* r-squared

```python
attrib = pd.DataFrame({f'GMWAX on {fac}': lfd(xsec2['GMWAX'], xsec2[fac]).rename({fac: 'Beta'}) 
                       for fac in ['SPY', 'GBI']}).T
attrib[['Alpha', 'Beta', 'R-squared', 't(Alpha)', 'Info Ratio']]
```

> | Regression | Alpha (ann.) | Beta | R² |
> |---|---|---|---|
> | GMWAX on SPY | 0.81% | 0.46 | 0.61 |

### 3. Interpretation

What do you conclude about `GMWAX` relative performance?
* Is it low-beta to the market?
* Does it track the index?
* Does `GMWAX` seem to earn alpha for its investors?

```python
#supporting numbers: GMWAX excess mean vs. what its beta alone would deliver, and downside capture
comp = pd.DataFrame(index = ['SPY', 'GBI'])
for fac in comp.index:
    b = attrib.loc[f'GMWAX on {fac}', 'Beta']
    comp.loc[fac, 'GMWAX mean'] = xsec2['GMWAX'].mean() * ann_factor
    comp.loc[fac, 'Beta x benchmark mean'] = b * xsec2[fac].mean() * ann_factor
    down = xsec2[fac] < 0
    comp.loc[fac, 'Down-week capture'] = xsec2.loc[down, 'GMWAX'].mean() / xsec2.loc[down, fac].mean()
    comp.loc[fac, 'Up-week capture'] = xsec2.loc[~down, 'GMWAX'].mean() / xsec2.loc[~down, fac].mean()
comp
```

> - **Low beta to the market?** Yes. The SPY beta is only 0.46, and GMWAX moves less than one-for-one even with its own benchmark (beta 0.81 to GBI). This matches its history of underweighting equities and holding a lot of fixed income and cash.
> - **Does it track the index?** Only loosely. R² is 0.74 against the GBI (correlation 0.86), so about a quarter of its variance is active, non-benchmark risk. The fund clearly deviates from its benchmark, as GMO's "conservative almost all the time, big bets on fat pitches" philosophy intends.
> - **Does it earn alpha?** Not meaningfully over the full sample. Alpha is +0.5% to +0.8%/yr (net of fees) with t-stats below 1. Most of GMWAX's 4.9% mean excess return is just its beta exposure: β × benchmark mean ≈ 4.1% (SPY) to 4.4% (GBI). Its up-week and down-week capture ratios are almost identical (~0.47–0.49 vs. SPY, ~0.83–0.85 vs. GBI). So the smaller drawdowns come from lower beta, not from asymmetric skill. Investors get roughly benchmark-like risk-adjusted returns with somewhat better tail outcomes, but no statistically reliable alpha.

### 1.

In `July 2015`, GMO forecasted the following as 7-year annualized mean **real** returns.

Their forecasts specify classes and benchmarks. Here, we map those to specific ETFs not specified by GMO. Additionally, we do not track every forecast made.

```python
forecasts = pd.DataFrame({
    'SPY': -2.0,
    'EFA': -0.5,
    'EEM': -3.2,
    'TLT': -0.9,
    'TIP': 0,
    'IJR': -2,
    'GWX': -1.1,
    'EMB': 2.1,
    'BWX': -2.8
},index=['forecast']).T

forecasts /= 100
forecasts.style.format('{:.1%}',na_rep='')
```

Use the `benchmarks` data to calculate the realized 7-year **nominal excess** return on these assets. You have ETF proxies for most of the categories, and ignore those for which you do not.

#### Note
Using nominal excess instead of real total should not greatly impact the analysis.

```python
# GMO's forecast made July 2015 -> 7-year window Aug 2015 through Jul 2022
fc_window = slice('2015-08-01', '2022-07-31')
xr = xrets.loc[fc_window, forecasts.index]
print(f'Realized window: {xr.index[0].date()} to {xr.index[-1].date()} ({len(xr)} weeks)')

years = len(xr) / ann_factor
r = rets.loc[fc_window, forecasts.index]
rf_win = rf.loc[fc_window]
compare = forecasts.copy()
compare['realized (arith. excess)'] = xr.mean() * ann_factor
compare['realized (geom. excess)'] = (1 + r).prod() ** (1 / years) - (1 + rf_win).prod() ** (1 / years)
compare['realized - forecast'] = compare['realized (geom. excess)'] - compare['forecast']
compare['forecast rank'] = compare['forecast'].rank(ascending=False)
compare['realized rank'] = compare['realized (geom. excess)'].rank(ascending=False)

display(compare.style.format({c: '{:.1%}' for c in compare.columns if 'rank' not in c} | {'forecast rank': '{:.0f}', 'realized rank': '{:.0f}'}))
print(f"Correlation (forecast, realized): {compare['forecast'].corr(compare['realized (geom. excess)']):.2f}")
print(f"Rank correlation: {compare['forecast'].corr(compare['realized (geom. excess)'], method='spearman'):.2f}")
```

>
> - **Level:** GMO's forecasts were far too pessimistic. Eight of nine assets beat their forecast, and the biggest misses were on **US large caps (+13 pp) and US small caps (+11 pp)**, the assets GMO considered most overvalued.

### 2.

* Given your out-of-sample results from the forecasting exercise, are you surprised by GMO's forecast accuracy?
* Could a dynamic strategy based on these signals have avoided the 2000–2011 equity underperformance?
* How does the in-sample vs. out-of-sample distinction relate to GMO's reliance on long-run mean reversion?

```python
ax = compare.plot.scatter(x='forecast', y='realized (geom. excess)', figsize=(7, 6),
                          title='GMO July 2015 7-yr forecast vs. realized (Aug 2015 - Jul 2022)')
for t, row in compare.iterrows():
    ax.annotate(t, (row['forecast'], row['realized (geom. excess)']), xytext=(4, 4), textcoords='offset points')
lims = [-0.06, 0.15]
ax.plot(lims, lims, ls='--', color='gray', lw=0.8)
ax.axhline(0, color='black', lw=0.5); ax.axvline(0, color='black', lw=0.5)
plt.show()
```

> **Are we surprised by GMO's forecast accuracy?** Not really. Valuation signals like D/P and E/P can look strong in-sample but tend to deliver little or even negative out-of-sample R² relative to a simple expanding-mean forecast. This is the expected result of the Chapter 7 exercise; check it against our own numbers once that notebook is done. GMO's 2001 forecast (Exhibit 1) looked remarkable, but it was one draw, made at an extreme valuation point, and the case itself notes that luck may have played a role. The July 2015 forecast was the opposite case: valuations still looked high, but margins and multiples did not revert, and US equities had one of their best 7-year runs. A track record built on a handful of independent 7-year periods cannot tell skill from luck with much confidence.
>
> **Could a dynamic strategy have avoided the 2000–2011 underperformance?** Partly. In 1999–2000 the dividend yield was at a record low and the CAPE was above 40, so any valuation-based strategy would have been heavily underweight equities going into the 2000–2002 crash. GMWAX did this and roughly doubled over 2001–2010 while SPY trailed T-bills. But the same signals had been bearish since about 1996–97, so the strategy would also have missed the late-1990s rally (as GMO did, losing 60% of its clients). It would have stayed underweight through much of the 2003–2007 and post-2009 recoveries as well. It avoids the worst losses but gives up a lot of upside, and its timing is poor.
>
> **In-sample vs. out-of-sample and GMO's reliance on mean reversion.** GMO's method needs "steady-state" levels for P/E (16, later 15) and profit margins (6%), and these are estimated from the same historical data the forecasts are judged on. That is effectively an in-sample assumption. If the long-run means shift (GMO's own "paradigm drift"), for example through structurally lower rates, higher margins from globalization and tech, or changes in buybacks and accounting, a mean-reversion forecast fails out of sample. That is what happened after 2012 and in the 2015 forecasts. GMO's revision of the steady-state P/E from 16 to 15 between 2007 and 2011 shows that these anchors get re-fit over time. Long-horizon regressions also use overlapping observations, so there are only about 20 independent 7-year periods since 1871. In-sample R² is inflated and real out-of-sample evidence is thin. Mean reversion may hold in the very long run, but its speed is uncertain, which is why GMO's results look strong in some periods (pre-2012) and weak in others (post-2012).
