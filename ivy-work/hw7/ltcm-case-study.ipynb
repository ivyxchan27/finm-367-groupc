### 1. 
Describe LTCM’s investment strategy with regard to the following aspects:
* Securities traded
* Trading frequency
* Skewness (Do they seek many small wins or a few big hits?)
* Forecasting (What is behind their selection of trades?)

**Answer:**
> | Aspect | LTCM's Investment Strategy |
> |---|---|
> | **Securities Traded** | LTCM primarily traded fixed-income securities and derivatives, including government bonds, corporate bonds, interest rate swaps, futures, and options, while also expanding into equity-related strategies. Its strategy centered on relative-value arbitrage, particularly convergence trading, by taking long positions in relatively undervalued securities and short positions in relatively overvalued securities. LTCM hedged common market exposures to isolate relative pricing discrepancies and employed substantial leverage to magnify returns. |
> | **Trading Frequency** | LTCM was an active but not high-frequency trader. Although it continuously monitored markets and adjusted positions as pricing discrepancies emerged, individual trades were generally held over medium- to long-term horizons, ranging from weeks to months or even years. Positions were typically maintained until the anticipated price convergence materialized. |
> | **Skewness** | LTCM sought many small wins rather than a few large gains, exposing the fund to negatively skewed returns. Under normal market conditions, convergence generated frequent, relatively small profits. However, during financial distress, pricing discrepancies could widen substantially rather than converge. Extensive leverage amplified these adverse movements, exposing LTCM to infrequent but potentially catastrophic losses. Consequently, attractive historical returns and Sharpe ratios could obscure significant downside tail risk. |
> | **Forecasting** | LTCM relied on quantitative financial models, historical statistical relationships, and economic theory to identify deviations from theoretically justified relative prices. Rather than forecasting the absolute direction of asset prices or broader financial markets, LTCM primarily forecasted the convergence of relative pricing discrepancies toward equilibrium. Its trade selection was therefore grounded in relative valuation and mean reversion. Profitability depended not only on eventual convergence but also on LTCM's ability to sustain its positions until convergence occurred. |

### 2. 
What are LTCM’s biggest advantages over its competitors?

**Answer:**
> LTCM's primary competitive advantages were its intellectual capital, sophisticated quantitative models, reputation, and superior access to financing. Its team combined academic expertise in financial economics with extensive practical trading experience, allowing it to identify relative pricing discrepancies and construct sophisticated hedged portfolios.
>
> More importantly, LTCM's reputation and relationships with major financial institutions enabled it to negotiate favorable financing terms, including low collateral requirements and access to substantial leverage. This allowed LTCM to profitably exploit small pricing discrepancies that might not have been economically viable for competitors facing greater financing constraints.
>
> LTCM also benefited from a substantial capital base, patient investor capital, and access to a broad range of financial markets. These advantages allowed it to pursue longer-term convergence trades and diversify across numerous investment opportunities. Thus, LTCM's competitive advantage stemmed not only from its ability to identify mispricing, but also from its ability to finance, implement, and sustain these trades.

### 3.
The case discusses four types of funding risk facing LTCM:
* collateral haircuts
* repo maturity
* equity redemption
* loan access

The case discusses specific ways in which LTCM manages each of these risks. Briefly discuss
them.

**Answer:**
> | Funding Risk | Description of Risk | LTCM's Risk Management Strategy |
> |---|---|---|
> | **Collateral Haircuts** | Lenders may increase collateral haircuts during periods of financial distress, requiring LTCM to commit additional working capital to maintain existing positions. | LTCM stress-tested worst-case haircuts (e.g., German government bonds: 0% to 2%; Italian government bonds: 0% to 5%; AAA-rated commercial mortgages: 2% to 10%). It forecasted potential liquidity requirements based on financing maturities and maintained substantial excess working capital. |
> | **Repo Maturity** | Counterparties may refuse to renew existing repo agreements, exposing LTCM to rollover risk and potentially forcing premature liquidation. | LTCM financed most securities positions through longer-term repos with maturities of 6–12 months, accepting less favorable financing rates in exchange for greater funding certainty. > | **Equity Redemption** | Simultaneous investor withdrawals may deplete the equity capital available to sustain investment positions. | LTCM initially imposed a three-year lock-up period. >

### 4. 

LTCM is largely in the business of selling liquidity and volatility. Describe how LTCM accounts
for liquidity risk in their quantitative measurements.

**Answer:**
> LTCM incorporated liquidity risk into its quantitative risk measurements by employing different volatility estimates and correlation structures over one-month and one-year horizons.
>
> - **Volatility:** LTCM assumed that short-term asset prices were influenced by both fundamental values and liquidity-driven trading pressures, whereas long-term prices were primarily determined by fundamental values. Consequently, its annualized short-term volatility estimates exceeded its long-term volatility estimates. For example, LTCM estimated a monthly standard deviation of 4 bps for the US Treasury swap-spread trade, equivalent to approximately 14 bps when annualized, compared to an estimated one-year standard deviation of 7 bps.
>
> - **Correlation:** LTCM assumed positive short-term correlations across trade categories because many financial institutions held similar positions and could be forced to liquidate simultaneously in response to liquidity pressures. Over longer horizons, LTCM based its correlations primarily on fundamental economic relationships and assumed that liquidity-induced correlations were positive but relatively small.
>
> Thus, LTCM explicitly incorporated liquidity risk into its short-term covariance estimates while treating liquidity-induced pricing discrepancies as temporary and potentially profitable investment opportunities over longer horizons.

### 5.

Is leverage risk currently a concern for LTCM?

**Answer:**
> From LTCM's perspective, leverage risk was not an immediate concern at the time of the case. As of August 1997, LTCM had $126.4 billion in assets and $6.7 billion in equity capital, implying an accounting leverage ratio of approximately 19:1. This was at the lower end of its historical range of 19:1-31:1 and comparable to major financial institutions.
>
> However, LTCM considered accounting leverage an inadequate measure of economic risk because it did not account for offsetting long-short positions or the underlying risk exposures of derivatives. Instead, LTCM evaluated risk through potential mark-to-market losses, portfolio volatility, and liquidity requirements. Its estimated annualized NAV volatility of approximately 11% was substantially below its target of 20%, and its estimated adverse liquidity requirements were well below its available financing capacity. Consequently, LTCM believed it was excessively capitalized rather than excessively leveraged.
>
> Nevertheless, leverage remained a significant underlying risk because this assessment depended on the accuracy of LTCM's volatility, correlation, and liquidity assumptions. Unexpected spread widening, increased correlations, or funding disruptions could generate substantial losses and force premature liquidation despite the portfolio's hedged structure.

### 6. 

Many strategies of LTCM rely on converging spreads. LTCM feels that these are almost win/win
situations because of the fact that if the spread converges, they make money. If it diverges, the
trade becomes even more attractive, as convergence is still expected at a future date.

What is the risk in these convergence trades?

**Answer:**
> The primary risk in LTCM's convergence trades is that pricing discrepancies may widen substantially before eventually converging, generating mark-to-market losses and increasing collateral requirements. Although divergence may make a trade more attractive from a fundamental valuation perspective, it simultaneously reduces LTCM's available capital and its ability to maintain or increase the position.
>
> Consequently, the profitability of convergence trading is path-dependent: even if LTCM correctly predicts eventual convergence, insufficient funding liquidity may force it to liquidate positions prematurely and realize substantial losses. This risk is amplified by leverage, limited market liquidity, and the possibility that multiple convergence trades diverge simultaneously due to common liquidity pressures.
>
> Furthermore, convergence is not necessarily guaranteed, as the underlying equilibrium relationship may change or LTCM's valuation models may be incorrect. Thus, convergence trading is not riskless arbitrage; its success depends on both the validity of the convergence forecast and LTCM's ability to remain solvent and sufficiently liquid until convergence materializes.

### 4. Nonlinear Exposure

Let's check for non-linear market exposure. Run the following regression on LTCM's **net** excess returns:

$$
\tilde{r}_t^{\text{ltcm}} = \alpha + \betalinear \tilde{r}_t^m + \betaquad \left(\tilde{r}_t^m\right)^2 + \epsilon_t

Report

* the linear and quadratic betas
* r-squared

```python
#quadratic market factor: squared SPY excess return
quad_factors = pd.DataFrame({"Linear": comparison_returns["SPY"], "Quadratic": comparison_returns["SPY"] ** 2})

ltcm_quad = hw8utils.lfd(comparison_returns["Net"], quad_factors)

ltcm_quad.to_frame("Net LTCM on SPY, SPY^2").T
```

**Answer:**

> | R-squared | 0.0207 | 0.0278 |
>
> **Nonlinear Market Exposure:** 
>> The negative quadratic beta (-2.1583) indicates a concave payoff profile, whereby large market movements in either direction negatively contribute to LTCM's returns relative to a linear payoff. 
>>
>> This is economically consistent with LTCM's strategy of selling liquidity and volatility, generating relatively frequent profits under stable market conditions while remaining exposed to extreme market movements.
>
> **Performance Attribution:**

>

### 6. 

Let's try to pinpoint the nature of LTCM's nonlinear exposure. Does it come more from exposure to up-markets or down-markets? Run the following regression on LTCM's net excess returns:

$$
\tilde{r}_t^{\text{ltcm}}  = \alpha + \beta\tilde{r}_t^m + \beta_u \max\left(\tilde{r}_t^m-k_1,0\right) + \beta_d \max\left(k_2 - \tilde{r}_t^m,0\right) + \epsilon_t
$$

where $k_1= .03$ and $k_2= -.03$. 

Report 
* annualized alpha
* market beta, the **up** and **down** betas
* r-squared

```python
#option-like market factors with strikes k1 = 3% (call-like, up-market) and k2 = -3% (put-like, down-market)
k1, k2 = 0.03, -0.03
spy = comparison_returns["SPY"]

option_factors = pd.DataFrame({"Market": spy, "Up": np.maximum(spy - k1, 0), "Down": np.maximum(k2 - spy, 0)})

ltcm_option = hw8utils.lfd(comparison_returns["Net"], option_factors)
display(ltcm_option.to_frame("Net LTCM on SPY, Up, Down").T)

#how many months each option-like factor is "in the money"
print(f"Months with SPY > {k1:.0%}: {(spy > k1).sum()}\n"
      f"Months with SPY < {k2:.0%}: {(spy < k2).sum()}\n"
      f"Total: {len(spy)}")
```

```python
#effective market beta in each region: the slope of LTCM on SPY implied by the option-factor regression
beta_m = ltcm_option["Beta Market"]
beta_u = ltcm_option["Beta Up"]
beta_d = ltcm_option["Beta Down"]

effective_betas = pd.DataFrame({"SPY Range": [f"< {k2:.0%}", f"{k2:.0%} to {k1:.0%}", f"> {k1:.0%}"],
                                "Slope": ["Market - Down", "Market", "Market + Up"],
                                "Effective Beta": [beta_m - beta_d, beta_m, beta_m + beta_u],
                                "Months": [(spy < k2).sum(), ((spy >= k2) & (spy <= k1)).sum(), (spy > k1).sum()]}, index = ["Down Market", "Normal Market", "Up Market"])

display(hw8utils.captioned(effective_betas, "Table 2.6: Effective market beta by SPY region, implied by the option-factor regression"))
```

```python
#fitted LTCM return implied by each model across a grid of SPY returns (monthly, not annualized)
spy_grid = np.linspace(spy.min(), spy.max(), 500)

fit_linear = ltcm_lfd["Alpha"] / 12 + ltcm_lfd["Beta SPY"] * spy_grid
fit_quadratic = ltcm_quad["Alpha"] / 12 + ltcm_quad["Beta Linear"] * spy_grid + ltcm_quad["Beta Quadratic"] * spy_grid ** 2
fit_option = (ltcm_option["Alpha"] / 12 + beta_m * spy_grid + beta_u * np.maximum(spy_grid - k1, 0) + beta_d * np.maximum(k2 - spy_grid, 0))

fig, ax = plt.subplots(figsize = (10, 6))
ax.scatter(spy, comparison_returns["Net"], alpha = 0.4, label = "Monthly returns")
ax.plot(spy_grid, fit_linear, linestyle = "--", label = "Linear model")
ax.plot(spy_grid, fit_quadratic, linestyle = "--", label = "Quadratic model")
ax.plot(spy_grid, fit_option, linewidth = 2.5, label = "Option-factor model")

#strikes of the option-like factors
ax.axvline(k1, color = "gray", linestyle = ":", label = f"Strikes ({k2:.0%}, {k1:.0%})")
ax.axvline(k2, color = "gray", linestyle = ":")
ax.axhline(0, color = "black", lw = 0.8)

ax.xaxis.set_major_formatter(plt.matplotlib.ticker.PercentFormatter(1))
ax.yaxis.set_major_formatter(plt.matplotlib.ticker.PercentFormatter(1))
ax.set_xlabel("SPY excess return")
ax.set_ylabel("Net LTCM excess return")
ax.set_title("Figure 2.6: Net LTCM vs. SPY monthly excess returns, linear, quadratic and option-factor fits")
ax.legend()
ax.grid(alpha = 0.2)
plt.tight_layout()
plt.show()
```
